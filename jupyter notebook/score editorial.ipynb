```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Load data
df = pd.read_csv('animes.csv')

# Preprocessing
df['Note_Globale'] = pd.to_numeric(df['Note_Globale'], errors='coerce')
df['Nb_Episodes'] = pd.to_numeric(df['Nb_Episodes'], errors='coerce')
df = df.dropna(subset=['Note_Globale', 'Nb_Episodes'])

# Definition of target 'bon'
# As discussed, threshold at 8.5 to create some separation
df["bon"] = (df["Note_Globale"] >= 8.5).astype(int)

# Features and target
X = df[["Note_Globale", "Nb_Episodes"]]
y = df["bon"]

# Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=42, 
    stratify=y
)

# Model
model = LogisticRegression(class_weight='balanced')
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Metrics
acc = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred, zero_division=0)

print(f"Accuracy: {acc}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report:")
print(class_report)

# Visualization of the class distribution
plt.figure(figsize=(8, 5))
sns.countplot(x='bon', data=df)
plt.title('Répartition des Animés (0: Moyen/Bon, 1: Excellent)')
plt.xlabel('Classe (Seuil 8.5)')
plt.ylabel('Nombre d\'animés')
plt.savefig('distribution_classes.png')

# Heatmap of correlation
plt.figure(figsize=(8, 6))
sns.heatmap(df[['Note_Globale', 'Nb_Episodes', 'bon']].corr(), annot=True, cmap='coolwarm')
plt.title('Matrice de Corrélation')
plt.savefig('correlation_machine_learning.png')
```

```python
import pandas as pd
from IPython.display import display

data = {
    "Anime": [
        "Naruto", "Trigum", "Dragon Ball Z",
        "One Piece", "Kuroko's Basket", "Blue Lock",
        "Naruto Shippuden", "GTO", "Bleach", "Tokyo Ghoul"
    ],
    "Note_Globale": [8.5, 8.3, 8.6, 7.9, 8.2, 9.1, 8.9, 9.0, 7.9, 7.5],
    "Note_Meilleur_Ep": [9.2, 9.0, 9.5, 8.8, 9.1, 9.8, 9.6, 9.7, 9.7, 9.6],
    "Note_Pire_Ep": [7.8, 7.5, 8.0, 7.0, 7.9, 8.5, 8.2, 8.4, 8.1, 8.0]
}

df = pd.DataFrame(data)
display(df)
```

```python
df["Ecart"] = df["Note_Meilleur_Ep"] - df["Note_Pire_Ep"]
df["Regularite"] = 10 - df["Ecart"]
df["Score_Metier"] = (df["Note_Globale"] * 0.7) + (df["Regularite"] * 0.3)
```

```python
import matplotlib.pyplot as plt

# compter les animés par statut
statuts = df["Status"].value_counts()

plt.figure(figsize=(8, 5))
couleurs = ["#1BCADE", "#C51BDE", "#DE1B38", "#1BDE4B"]

plt.bar(statuts.index, statuts.values, color=couleurs)
plt.title("Répartition des animés par statut")
plt.xlabel("Statut")
plt.ylabel("Nombre d'animés")

plt.show()
```

```python
df["Anime"] = df["Anime"].str.upper()

df_trie = df.sort_values("Score_Metier", ascending=False)

print("TOP 10 ANIMES:")
print(df_trie[['Anime', 'Note_Globale', 'Regularite', 'Score_Metier']]
      .head(10)
      .to_string(index=False))

print("\nColonnes disponibles:")
print("-" * 40)
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")
```
